Excel is useful for small analyses, but these limitations can make Python a better fit for larger, repeatable engineering workflows:

Q.2. Write a Python function that converts temperature from Fahrenheit to Celsius. Include a docstring, type hints, and input validation. Test it with three values including one invalid input.

```python
def fahrenheit_to_celsius(temp_f: float) -> float:
    """Convert a temperature from Fahrenheit to Celsius.

    Args:
        temp_f: Temperature in degrees Fahrenheit.

    Returns:
        Temperature in degrees Celsius.

    Raises:
        TypeError: If temp_f is not a number.
    """
    if isinstance(temp_f, bool) or not isinstance(temp_f, (int, float)):
        raise TypeError("Temperature must be a number.")

    return (temp_f - 32) * 5 / 9

print(fahrenheit_to_celsius(32))
print(fahrenheit_to_celsius(212))

try:
    print(fahrenheit_to_celsius("hot"))
except TypeError as error:
    print(f"Invalid input: {error}")
```

The formula is °C = (°F − 32) × 5/9. This function accepts numbers, rejects non-numeric input, and does not treat True or False as numbers.

Q.6. Write a Python class called SubseaPipeline with properties for length_km, od_inches, wt_mm, material_grade, and a method that calculates the pipe volume in cubic metres.

```python
import math


class SubseaPipeline:
    """Represent a pipeline and calculate its internal volume."""

    def __init__(
        self,
        length_km: float,
        od_inches: float,
        wt_mm: float,
        material_grade: str,
    ) -> None:
        """Store pipeline dimensions and material grade."""
        if length_km <= 0 or od_inches <= 0 or wt_mm <= 0:
            raise ValueError("Length, outside diameter, and wall thickness must be positive.")

        self.length_km = length_km
        self.od_inches = od_inches
        self.wt_mm = wt_mm
        self.material_grade = material_grade

    def pipe_volume_m3(self) -> float:
        """Return the pipe's internal capacity in cubic metres."""
        od_m = self.od_inches * 0.0254
        wall_thickness_m = self.wt_mm / 1000
        inside_diameter_m = od_m - 2 * wall_thickness_m

        if inside_diameter_m <= 0:
            raise ValueError("Wall thickness leaves no positive internal diameter.")

        length_m = self.length_km * 1000
        radius_m = inside_diameter_m / 2

        return math.pi * radius_m**2 * length_m


pipeline = SubseaPipeline(
    length_km=10,
    od_inches=12,
    wt_mm=20,
    material_grade="X65",
)

print(pipeline.pipe_volume_m3())
```

This example interprets “pipe volume” as the internal capacity of the pipe. It uses outside diameter minus twice the wall thickness to find the inside diameter, then calculates the cylindrical volume.

Q.7. Why is version control essential for engineering software? Describe three specific scenarios where Git would prevent costly errors in an offshore engineering context.

Git records changes to code, lets a team review them, and can restore an earlier version. For example:

1. A conversion factor is changed incorrectly: A reviewer can see the change and restore the approved factor before it affects engineering calculations.

2. A Barlow validation rule is removed by mistake: The team can identify when the check disappeared and recover the version that rejected invalid pipe dimensions.

3. A new asset-data import changes earlier results: Git can show which code change caused the difference and let engineers compare it with the previous calculation workflow.

Q.9. Explain the try/except pattern. Write a function that safely divides two numbers and returns ‘Error: Division by zero’ if the denominator is zero.

The try/except pattern is used in Python to handle errors without making the whole program crash. For Example; TRY to do something. if an ERROR happens, EXCEPT it and do something else.

```python
#for instance;
try:
    #code that might cause an error

except:
    #what to do if an error occurs

#For example:
try:
    number = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero.")
```

Instead of the program crashing, Python catches the ZeroDivisionError and prints the message. Below is the fumction to safely divide two numbers

```python
def safe_divide(numerator: float, denominator: float) -> float | str:
    """Divide two numbers, returning an error message for division by zero."""
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return "Error: Division by zero"


print(safe_divide(10, 2))
print(safe_divide(10, 0))
```
